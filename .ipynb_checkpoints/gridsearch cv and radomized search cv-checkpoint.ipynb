```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
```

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
```

```python
df = pd.read_csv("diabetes.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df['Outcome'].value_counts()
```

```python
X = df.drop(columns = ["Outcome"] , axis=1)
Y = df["Outcome"]
```

```python
print(Y)
```

```python
print(X)
```

```python
X.std()
```

```python
scaler = StandardScaler()
scaler.fit(X)
stand_x = scaler.transform(X)
```

```python
stand_x.std()
```

```python
X = np.asarray(stand_x)
Y = np.asarray(Y)
```

```python
grid = {
   
    'C': [1,5,10,20],
    'kernel': ['rbf', 'linear', 'sigmoid', 'poly']
}
```

```python
model = SVC()
classifier = GridSearchCV(model ,grid , cv =5)
```

```python
classifier.fit(X,Y)
```

```python
classifier.cv_results_
```

```python
print("best parametrs = ",classifier.best_params_ )
```

```python
print("highest accuracy  = ", classifier.best_score_)
```

```python
df_ = pd.DataFrame(classifier.cv_results_)
```

```python
df_.head()
```

```python

grid_search_results = df_[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
grid_search_results
```

```python
classifier = RandomizedSearchCV(model ,grid , cv =5)
```

```python
classifier.fit(X,Y)
```

```python
df2 = pd.DataFrame(classifier.cv_results_)
```

```python
random_search_results = df2[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
random_search_results
```

```python
print("best parametrs = ",classifier.best_params_ )
```
